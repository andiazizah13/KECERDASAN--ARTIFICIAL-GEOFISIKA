# **Prompt AI untuk Metode Geofisika**

Pada tugas ini membuat prompt AI dengan menerapkan struktur penyusunan prompt yang terdiri dari **Role, Task, Context, Input, Constraints, Output, Quality Criteria,** dan **Verification**.

Adapun metode yang saya ambil yaitu :
1. Electrical Resistivity Tomography - ERT
2. Seismic Reflection
3. Ground Penetrating Radar (GPR)
4. MASW (Multichannel Analysis of Surface Waves)
5. HVSR (Horizontal to Vertical Spectral Ratio / Mikrotremor)

--------

# **1. (Electrical Resistivity Tomography - ERT)**
R — Role Bertindak sebagai seorang geofisikawan ahli eksplorasi air tanah dan lingkungan (near-surface geophysicist) yang berpengalaman dalam menganalisis penampang resistivitas.

T — Task Analisis data penampang resistivitas (pseudosection 2D) untuk mengidentifikasi keberadaan lapisan akuifer air tawar dan memetakan zona yang terindikasi mengalami intrusi air laut.

C — Context Survei geolistrik tahanan jenis dilakukan di daerah pesisir pantai yang padat penduduk. Terdapat kekhawatiran dari warga bahwa sumur-sumur dangkal mereka mulai terasa payau akibat intrusi air laut ke dalam akuifer dangkal.

I — Input Data yang diberikan berupa:

Gambar penampang model inversi resistivitas 2D (dalam satuan Ohm.meter).
Data topografi lintasan pengukuran.
Informasi geologi regional mengenai litologi penyusun (misal: batu pasir, lempung, dan batu gamping).
Data log sumur bor terdekat (jika tersedia).
K — Constraints Jangan menyimpulkan jenis batuan secara definitif hanya berdasarkan satu nilai resistivitas saja, ingat prinsip ambiguitas (equivalence problem). Bedakan secara cermat rentang nilai resistivitas rendah yang diakibatkan oleh intrusi air asin dengan nilai rendah akibat lapisan lempung murni.

O — Output Berikan:

Evaluasi sebaran nilai resistivitas (zona konduktif vs zona resistif).
Identifikasi kedalaman dan ketebalan lapisan yang diprediksi sebagai akuifer.
Delineasi zona anomali resistivitas sangat rendah yang diinterpretasikan sebagai intrusi air laut.
Korelasi antara anomali dengan ketersediaan data geologi regional dan sumur bor.
Kesimpulan kelayakan area untuk pengeboran air tawar baru.
Q — Quality Criteria Interpretasi harus mematuhi tabel rentang nilai resistivitas batuan dan fluida standar, serta secara logis menyertakan konsep intrusi baji air asin (saltwater wedge) pada wilayah pesisir.

V — Verification Periksa kembali apakah kedalaman zona konduktif yang diinterpretasikan sebagai intrusi air laut berkorelasi positif dengan letak muka air tanah dan salinitas pada data sumur observasi yang ada.

## **2. Seismik Refleksi (Seismic Reflection)**
R — Role Bertindak sebagai ahli geofisika eksplorasi migas (Seismic Interpreter) senior yang terampil dalam analisis fasies seismik dan struktur bawah permukaan.

T — Task Lakukan penelusuran (picking) horizon utama dan identifikasi struktur patahan pada penampang seismik untuk mencari prospek jebakan hidrokarbon (hydrocarbon trap).

C — Context Data seismik refleksi ini diambil dari cekungan sedimen lepas pantai (offshore) yang secara regional diketahui memiliki sistem petroleum aktif dengan potensi perangkap struktural berupa antiklin dan sesar normal.

I — Input Data yang diberikan berupa:

Gambar penampang seismik 2D Pre-Stack Time Migration (PSTM).
Penanda kedalaman sumur (well-to-seismic tie) jika ada.
Log sonik dan densitas dari sumur eksplorasi terdekat.
K — Constraints Hindari menginterpretasikan multiple, diffraction, atau artefak migrasi sebagai horizon atau struktur geologi asli. Jangan mengasumsikan setiap amplitudo tinggi (bright spot) pasti berisi hidrokarbon tanpa mempertimbangkan efek tuning (tuning effect) atau litologi batubara.

O — Output Berikan:

Deskripsi karakteristik reflektor (amplitudo, kontinuitas, dan frekuensi).
Identifikasi patahan utama beserta jenis sesarnya (naik, turun, mendatar).
Analisis jebakan struktural (antiklin/sesar) yang potensial menjadi reservoir.
Identifikasi anomali amplitudo yang berpotensi menjadi Direct Hydrocarbon Indicator (DHI).
Ringkasan prospek interpretasi struktural dan stratigrafi.
Q — Quality Criteria Horizon yang diinterpretasikan harus konsisten secara seismik stratigrafi, dan patahan yang digambar harus memiliki pergeseran (throw) yang logis secara mekanika batuan regional.

V — Verification Lakukan validasi silang (cross-validation) penampang hasil interpretasi dengan synthetic seismogram dari data sumur terdekat untuk memastikan bahwa fase seismik telah di-tie dengan benar.

## **3.Ground Penetrating Radar (GPR)**
R — Role Bertindak sebagai ahli geofisika dangkal (near-surface geophysicist) dan ahli geoteknik yang ahli dalam mendeteksi utilitas infrastruktur terpendam.

T — Task Interpretasi profil radargram B-scan untuk mendeteksi utilitas bawah permukaan (pipa, kabel) dan rongga (sinkhole) sebelum dilakukan pekerjaan penggalian berat.

C — Context Survei GPR difokuskan pada area jalan protokol perkotaan yang akan dibangun pondasi pilar jalur kereta api layang (LRT). Sangat penting untuk menghindari kerusakan pada pipa air utama dan kabel serat optik saat pemancangan.

I — Input Data yang diberikan berupa:

Profil radargram B-scan pasca-processing (sudah melalui dewow, gain, filtering).
Data asumsi kecepatan gelombang elektromagnetik (v) pada material jalan aspal dan tanah urugan bawahnya.
Frekuensi antena yang digunakan (misal: 250 MHz atau 500 MHz).
K — Constraints Jangan menginterpretasikan ringing effect atau artefak pita frekuensi tinggi (noise) sebagai struktur utilitas. Ingat bahwa pada tanah dengan kadar lempung tinggi (konduktif), kedalaman penetrasi sinyal radar sangat terbatas.

O — Output Berikan:

Identifikasi bentuk anomali hiperbolik yang mencerminkan objek titik/silindris (pipa/kabel).
Estimasi posisi (jarak pada lintasan) dan kedalaman (dalam meter) dari anomali hiperbola tersebut.
Identifikasi zona anomali amplitudo tinggi ireguler yang berpotensi sebagai rongga/rongga aspal (voids).
Ringkasan mitigasi risiko penggalian di jalur tersebut.
Q — Quality Criteria Perhitungan kedalaman anomali harus menggunakan estimasi Two-Way Travel Time (TWT) dan konstanta dielektrik (relative permittivity) yang relevan dengan jenis aspal/tanah di area observasi.

V — Verification Bandingkan estimasi kedalaman utilitas hasil radargram dengan denah (as-built drawing) pipa milik PDAM lokal. Jika ada keraguan, rekomendasikan penggalian uji dangkal (test pit).

## **4.  MASW (Multichannel Analysis of Surface Waves)**
R — Role Bertindak sebagai seorang ahli geofisika dangkal (near-surface geophysicist) dan ahli geoteknik yang berspesialisasi dalam analisis gelombang permukaan untuk keperluan rekayasa sipil.

T — Task Lakukan analisis ekstraksi kurva dispersi gelombang Rayleigh dan inversi model untuk menghasilkan profil kecepatan gelombang geser (Vs) 1D/2D bawah permukaan.

C — Context Survei seismik MASW dilakukan di sebuah lahan kosong yang akan dibangun gedung bertingkat tinggi. Tujuannya adalah untuk mengetahui daya dukung tanah, mengidentifikasi kedalaman batuan dasar (bedrock), dan menentukan kelas situs seismik (Site Class) sebagai syarat perancangan bangunan tahan gempa.

I — Input Data yang diberikan berupa:

Rekaman seismogram (gelombang permukaan/Rayleigh) dari tembakan palu godam (sledgehammer).
Data geometri survei (spasi antar geophone, offset tembakan).
Kurva overtone image (spektrum kecepatan fase vs frekuensi) pasca-transformasi domain.
Data pengeboran geoteknik awal (jika ada).
K — Constraints Jangan memaksakan penarikan (picking) kurva dispersi pada mode fundamental (fundamental mode) jika sinyal didominasi oleh mode yang lebih tinggi (higher modes) atau tertutup oleh noise spasial (seperti aliasing). Hati-hati bahwa resolusi lapisan dalam (frekuensi rendah) sangat bergantung pada spread length (panjang lintasan).

O — Output Berikan:

Kurva dispersi teoritis vs observasi hasil penarikan (picking).
Penampang profil kecepatan gelombang geser (Vs) terhadap kedalaman (model 1D atau 2D).
Perhitungan nilai rata-rata kecepatan gelombang geser hingga kedalaman 30 meter (Vs30).
Penentuan kedalaman lapisan keras atau batuan dasar (engineering bedrock).
Klasifikasi jenis tanah (tanah lunak, sedang, atau batuan keras) berdasarkan pedoman SNI Gempa.
Q — Quality Criteria Model kecepatan (Vs) hasil inversi harus menghasilkan nilai Root Mean Square (RMS) error yang kecil terhadap kurva dispersi observasi, dan profil lapisannya tidak boleh memiliki lonjakan nilai kecepatan yang sangat drastis secara tidak logis tanpa dasar geologi/geoteknik.

V — Verification Bandingkan profil kecepatan gelombang geser (Vs) yang dihasilkan dengan nilai pengujian penetrasi standar (N-SPT) dari log sumur bor geoteknik terdekat. Nilai Vs tinggi harus berkorelasi dengan nilai N-SPT yang tinggi (lapisan keras).

## **5. HVSR (Horizontal to Vertical Spectral Ratio / Mikrotremor)**
R — Role Bertindak sebagai seismolog rekayasa dan ahli mikrozonalisasi gempa yang berpengalaman dalam menganalisis data getaran mikrotremor ambien.

T — Task Analisis data rekaman mikrotremor (ambient noise) menggunakan metode spektral rasio (H/V) untuk menentukan parameter dinamis tanah seperti frekuensi natural (f0) dan faktor amplifikasi (A0).

C — Context Survei mikrotremor single-station dilakukan secara grid di kawasan perkotaan yang pernah terdampak gempa bumi. Survei ini bertujuan untuk memetakan zona kerentanan seismik (seismic vulnerability) guna mitigasi bahaya resonansi pada infrastruktur dan perumahan warga di masa depan.

I — Input Data yang diberikan berupa:

Rekaman seismogram mikrotremor 3 komponen (2 komponen horizontal N-S dan E-W, serta 1 komponen vertikal Z).
Durasi data rekaman (minimal 30 menit per titik).
Peta geologi regional dan jenis batuan sedimen di wilayah survei.
K — Constraints Data yang diolah harus terbebas dari transient noise lokal jarak dekat (seperti mesin bergetar tiba-tiba, langkah kaki tepat di sebelah sensor, atau tetesan hujan besar). Jangan menganalisis window sinyal yang secara jelas mengandung lonjakan anomali (spikes) akibat gangguan buatan yang sangat dekat.

O — Output Berikan:

Kurva Spektral Rasio H/V yang menunjukkan rasio spektrum horizontal terhadap vertikal.
Nilai estimasi frekuensi natural/dominan tanah (f0) dan faktor amplifikasi (A0).
Perhitungan Indeks Kerentanan Seismik (Kg) di titik ukur tersebut.
Estimasi ketebalan sedimen lunak hingga batuan dasar (bedrock) jika nilai kecepatan gelombang geser (Vs) lokal diketahui.
Q — Quality Criteria Bentuk kurva H/V yang dihasilkan harus memenuhi pedoman reliabilitas SESAME (Eropa), yaitu memiliki puncak amplitudo (peak) yang jelas, kurva berbentuk cukup sempit (tidak landai rata), serta konsisten/stabil saat diuji melintasi berbagai pemotongan window waktu.

V — Verification Korelasikan frekuensi natural tanah (f0) yang didapat dengan pola sebaran kerusakan bangunan saat gempa historis terjadi (bangunan yang rusak paling parah biasanya memiliki frekuensi natural yang sama dengan f0 tanah akibat efek resonansi).
